## 2. Product Distribution

The distribution of orders across products was analyzed to determine which products appeared most frequently in the dataset.

The Printer was the most frequently ordered product with 181 orders, while Phone had the lowest number of orders with 156. Tablet, Chair, Laptop, Desk, and Monitor recorded 179, 178, 173, 170, and 163 orders respectively.

Overall, the distribution across the seven product categories is relatively balanced, with only a 25-order difference between the most and least frequently ordered products.

```python
df.groupby("Product")["TotalPrice"].sum().sort_values(ascending=False)
```

Product Sales Performance: Chairs generated the highest total sales value at ₦195,620.11, closely followed by Printers at ₦195,612.61. Phones recorded the lowest total sales value at ₦151,722.39. This indicates that Chairs and Printers were the strongest contributors to overall sales revenue, while Phones contributed comparatively less.

```python
product_sales = df.groupby('Product')['TotalPrice'].sum().sort_values(ascending=False)

fig = px.bar(
    x=product_sales.index,
    y=product_sales.values,
    title='Total Sales by Product',
    labels={
        'x': 'Product',
        'y': 'Total Sales'
    }
)

fig.show()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
monthly_sales = df.groupby(
    df['Date'].dt.to_period('M')
)['TotalPrice'].sum()

monthly_sales
```

### Sales Trend Analysis 
This analysis examines how total sales changed over time to identify noticeable trends, increases, decreases, and fluctuations in sales.

```python
monthly_sales = monthly_sales.reset_index()

monthly_sales['Date'] = monthly_sales['Date'].dt.to_timestamp()
```

```python
fig = px.line(
    monthly_sales,
    x='Date',
    y='TotalPrice',
    title='Total Sales Over Time',
    labels={
        'Date': 'Date',
        'TotalPrice': 'Total Sales'
    }
)

fig.show()
```

### Observation

The line graph shows how total sales changed over time. It helps identify periods with higher or lower sales and makes it easier to observe any noticeable trends or fluctuations in sales throughout the dataset.

```python
order_status = df['OrderStatus'].value_counts()

fig = px.bar(
    x=order_status.index,
    y=order_status.values,
    title='Order Status Distribution',
    labels={
        'x': 'Order Status',
        'y': 'Number of Orders'
    }
)

fig.show()
```

### Observation

The bar chart shows the distribution of orders across the different order statuses. Cancelled and Returned orders have the highest counts, while Delivered orders have the lowest count among the five statuses.

Outlier Analysis

This analysis examines the distribution of total prices to identify unusually high or low order values that may represent potential outliers.

```python
fig = px.box(
    df,
    y='TotalPrice',
    title='Distribution and Outliers of Total Price',
    labels={
        'TotalPrice': 'Total Price'
    }
)

fig.show()
```

### Observation

The box plot shows the distribution of total prices and helps identify potential outliers. Values appearing beyond the whiskers represent unusually high or low order totals that may require further investigation.
