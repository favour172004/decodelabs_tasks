## 1. Handling Missing Values

The dataset was checked for missing values using `isnull().sum()`.

The `CouponCode` column contained 309 missing values. These missing values were investigated using `value_counts()` to understand whether they represented an actual data issue.

Since the `CouponCode` field is optional, the missing values were interpreted as orders where no coupon was used. Therefore, the missing values were replaced with `"No Coupon"` rather than removing the affected records.

After cleaning, the `CouponCode` column contained four categories: `FREESHIP`, `WINTER15`, `SAVE10`, and `No Coupon`.

## 4. Cleaning and Validation of Numerical Columns

The numerical columns `Quantity`, `UnitPrice`, and `TotalPrice` were inspected for missing, zero, negative, and inconsistent values.

The `Quantity` column contains values ranging from 1 to 5, with no zero or negative quantities identified. The `UnitPrice` column contains values ranging from 11.39 to 699.93, with no zero, negative, or missing prices found.

The `TotalPrice` column was validated by comparing it with the calculated value of `Quantity × UnitPrice`. The largest difference across all 1,200 records was approximately 4.55 × 10⁻¹³, which is effectively zero and is attributed to floating-point precision.

Therefore, no corrections were required for the numerical columns.
